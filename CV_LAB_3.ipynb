```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def average_filter(img, k):
    p = 0
    s = 1
    kernel = np.ones((k, k)) * (1 / (k * k))
    m, n = img.shape

    filtered_img = np.zeros((m - k + 1, n - k + 1))

    for i in range(0, m - k + 1, s):
        for j in range(0, n - k + 1, s):

            filtered_img[i, j] = np.sum(
                kernel * img[i:i+k, j:j+k]
            )

    return filtered_img



img = Image.open("/content/IMAGE2.jpg")

gray_img = img.convert("L")
img_array = np.array(gray_img)
blur_3 = average_filter(img_array, 3)
blur_5 = average_filter(img_array, 5)
blur_7 = average_filter(img_array, 7)
blur_11 = average_filter(img_array, 11)
plt.figure(figsize=(12, 10))
plt.subplot(2, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.subplot(2, 3, 2)
plt.imshow(img_array, cmap="gray")
plt.title("Gray Scale Image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(blur_3, cmap="gray")
plt.title("Average Filter 3x3")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(blur_5, cmap="gray")
plt.title("Average Filter 5x5")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(blur_7, cmap="gray")
plt.title("Average Filter 7x7")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(blur_11, cmap="gray")
plt.title("Average Filter 11x11")
plt.axis("off")

plt.tight_layout()
plt.show()
```
