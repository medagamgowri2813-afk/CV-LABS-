```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def separate_lsb(image):
    lsb_plane = (image >> 0) & 1
    lsb_plane = lsb_plane * 255
    return lsb_plane.astype(np.uint8)

img = cv2.imread("/content/hqdefault.jpg", cv2.IMREAD_GRAYSCALE)

lsb_image = separate_lsb(img)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(lsb_image, cmap='gray')
plt.title('LSB Plane (Bit Plane 0)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
